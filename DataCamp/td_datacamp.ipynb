```python
import os
import pandas as pd
import s3fs
import sklearn
import matplotlib.pyplot as plt
```

```python
os.environ['AWS_S3_ENDPOINT']
S3_ENDPOINT_URL = 'http://' + os.environ['AWS_S3_ENDPOINT']
fs = s3fs.S3FileSystem(client_kwargs = {'endpoint_url' :S3_ENDPOINT_URL })
fs.ls('mlepennec-ensae')
BUCKET = 'mlepennec-ensae'
FILE_KEY_S3 = '/engieX.csv'
FILE_PATH_S3 = BUCKET+FILE_KEY_S3
with fs.open(FILE_PATH_S3, mode = "rb") as file_in : 
    X_df = pd.read_csv(file_in,delimiter=';')
```

```python
FILE_KEY_S3 = '/engieY.csv'
FILE_PATH_S3 = BUCKET+FILE_KEY_S3
with fs.open(FILE_PATH_S3, mode = "rb") as file_in : 
    y_df = pd.read_csv(file_in,delimiter=';')
```

```python
df = X_df.merge(y_df, left_on='ID', right_on='ID')
```

```python
WT1_df = df[df.MAC_CODE == 'WT1']
WT1_df
```

```python
WT1_df.columns
```

```python
WT1_df = WT1_df.drop(columns=[col for col in WT1_df.columns if col.endswith("min") or col.endswith("max") or col.endswith('std')])
```

```python
WT1_df = WT1_df[WT1_df.Date_time % 6 == 1]
WT1_df
```

```python
for col in WT1_df.columns:
    n_na=WT1_df[col].isna().sum()
    if n_na > 0:
        print(f"{col} : {n_na}")
```

```python
WT1_df = WT1_df.drop(columns=["Generator_converter_speed", "Gearbox_inlet_temperature", "Grid_voltage", "Absolute_wind_direction_c", "Nacelle_angle_c", "Date_time", 'ID', 'MAC_CODE'])
```

```python
WT1_df = WT1_df.reset_index(drop=True)
WT1_df
```

```python
cv_thres = 0.8*len(WT1_df)
X = WT1_df.drop(columns = ['TARGET'])
Y = WT1_df['TARGET']
X_train = X[X.index < cv_thres].copy()
X_test = X[X.index >= cv_thres].copy()
Y_train = Y[Y.index < cv_thres].copy()
Y_test = Y[Y.index >= cv_thres].copy()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import RidgeCV, ElasticNetCV, LassoCV
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import Pipeline
from sklearn.model_selection import KFold
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error
```

```python
simple_lm_model = LinearRegression().fit(X_train, Y_train)

# Erreur d'entraînement
print(mean_absolute_error(simple_lm_model.predict(X_train), Y_train))

# Erreur de test
print(mean_absolute_error(simple_lm_model.predict(X_test), Y_test))
```

```python
random_forest = RandomForestRegressor(n_estimators=100, random_state=42).fit(X_train, Y_train)

print(mean_absolute_error(random_forest.predict(X_train), Y_train))

print(mean_absolute_error(random_forest.predict(X_test), Y_test))
```

```python
grad_boost = GradientBoostingRegressor(random_state=42).fit(X_train, Y_train)

print(mean_absolute_error(grad_boost.predict(X_train), Y_train))

print(mean_absolute_error(grad_boost.predict(X_test), Y_test))
```
