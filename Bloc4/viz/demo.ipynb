```python
notes = {
"Nom": ["Teddy", "Jenny", "Cindy", "Franky", "Jimmy"],
"Note": [12, 18, 14, 12, 16],
}

pn.panel(notes).servable()
```

```python
import pandas as pd
df = pd.DataFrame(notes)

pn.panel(df).servable()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots()
ax.bar(df.Nom, df.Note)
ax.grid(linestyle=":")

plt.close(fig) # Important pour éviter les fuites de mémoire

pn.panel(fig, width=512).servable()
```

```python
import plotly.express as px

pn.extension("plotly") # Nécessaire pour Plotly

fig_plotly = px.line(df, x="Nom", y="Note")
fig_plotly.update_traces(mode="lines+markers")
fig_plotly.layout.autosize = True

pn.panel(fig_plotly).servable()
```
